```python
# Save as student_report.py
import sys

name = sys.argv[1]
mark1 = int(sys.argv[2])
mark2 = int(sys.argv[3])
mark3 = int(sys.argv[4])

total = mark1 + mark2 + mark3
average = total / 3

print("Student name:", name)
print("Total marks:", total)
print("Average marks:", average)
```

```python
marks = {"Maths": 80, "Science": 90, "English": 70}
total = 0

for value in marks.values():
    total = total + value

print("Sum of dictionary values:", total)
```

```python
numbers = [1, 2, 2, 3, 3, 3, 4]
duplicates = 0
checked = []

for num in numbers:
    if numbers.count(num) > 1 and num not in checked:
        duplicates = duplicates + 1
        checked.append(num)

print("Number of duplicate elements:", duplicates)
```
