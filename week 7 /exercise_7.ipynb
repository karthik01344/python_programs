```python
squares = { }

for i in range(1, 6):
    squares[i] = i * i

print(squares)
```

```python
numbers = [1, 2, 2, 3, 1, 2]
frequency = { }

for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num] + 1
    else:
        frequency[num] = 1

print(frequency)
```

```python
# Write mode
file = open("sample.txt", "w")
file.write("Hello Python\n")
file.close()

# Append mode
file = open("sample.txt", "a")
file.write("Welcome to file handling")
file.close()

# Read mode
file = open("sample.txt", "r")
print(file.read())
file.close()
```

```python
file = open("sample.txt", "r")
lines = file.readlines()
file.close()

for line in lines[::-1]:
    print(line.strip())
```

```python
file = open("sample.txt", "r")
text = file.read()
file.close()

characters = len(text)
words = len(text.split())
lines = len(text.splitlines())

print("Characters:", characters)
print("Words:", words)
print("Lines:", lines)
```
